Simple solution of the Blasius boundary layer equations

$$f^{\prime\prime\prime}+\frac 12 f f^\prime = 0$$

subject to the boundary conditions

$$u(x,y=0)=\left.\frac{df}{d\eta}\right|_{\eta=0}=0$$  
$$v(x,y=0)=f(\eta=0)=0$$  
$$u(x,y\rightarrow \infty)=\left.\frac{df}{d\eta}\right|_{\eta=\infty}=1$$  

```python
# import required libs
import numpy as np
from scipy.integrate import odeint
from pylab import *
from IPython.display import Latex
import pandas as pd
```

The present numerical solution uses a simple a shooting method.

First, the differential equations is assemble as a first order system using

$$\mathbf{f}
=
\begin{pmatrix}
f\\
f^\prime\\
f^{\prime\prime}
\end{pmatrix}
=
\begin{pmatrix}
f_0\\
f_1\\
f_2
\end{pmatrix}
$$

giving

$$
\frac{d}{d \eta}
\begin{pmatrix}
f\\
f^\prime\\
f^{\prime\prime}
\end{pmatrix}
=
\frac{d}{d \eta}
\begin{pmatrix}
f_0\\
f_1\\
f_2
\end{pmatrix}
=
\begin{pmatrix}
f_1\\
f_2\\
-\frac 12 f_0 f_2
\end{pmatrix}
$$

subject to the boundary conditions

$\displaystyle u(x,y=0)=f_1=0$  
$\displaystyle v(x,y=0)=f_0=0$

and $\displaystyle \left.\frac{d^2f}{d\eta^2}\right|_{\eta=0}$ is changed iteratively until $f_1(\eta=\eta_\max)=1$  

where $\eta_\max$ is a large constant, say 12


```python
def Blasius( f, η ):
    return ( f[1], f[2], -0.5 * f[0] * f[2] )
```

```python
ηmax = 8
η = np.linspace( 0, ηmax, ηmax*10+1 )
```

```python
f_initial = ( 0.0, 0.0, 0.3 )
solution = odeint( Blasius, f_initial, η )
f = solution.transpose()

plot( f[0], η, label='$f$' )
plot( f[1], η, label='$f\'$' )
plot( f[2], η, label='$f\'\'$' )
ylabel( r'$\eta$')
legend()
grid()
xlim( (-0.1,4) )
ylim( (0,8) );
```

```python
f_initial = ( 0.0, 0.0, 0.35 )
solution = odeint( Blasius, f_initial, η )
f = solution.transpose()

plot( f[0], η, label='$f$' )
plot( f[1], η, label='$f\'$' )
plot( f[2], η, label='$f\'\'$' )
ylabel( r'$\eta$')
legend()
grid()
xlim( (-0.1,4) )
ylim( (0,8) );
```

```python
f2_Boyd = 0.33205733621519630
f_initial = ( 0.0, 0.0, f2_Boyd )
solution = odeint( Blasius, f_initial, η )
f = solution.transpose()

plot( f[0], η, label='$f$' )
plot( f[1], η, label='$f\'$' )
plot( f[2], η, label='$f\'\'$' )
ylabel( r'$\eta$')
legend()
grid()
xlim( (-0.1,4) )
ylim( (0,8) );
Latex( r'$f^\prime=\left(df/d\eta\right)|_{\eta_\max}=%.14f$' % f[1][-1] )
```

Plot of the dimensionless velocity $\displaystyle \frac{u}{U_e}=f_1$

```python
plot( f[1], η )
xlabel( r'$\frac{u}{U_e}$' )
ylabel( r'$\eta$' )
grid()
xlim( (-0.1,1.1) )
ylim( (0,8) );
```

Plot of the dimensionless velocity $\displaystyle \frac{v}{U_e}\sqrt{\text{Re}_x}=\frac 12 \left( \eta f_1 - f_0\right)$

```python
v = 0.5 * ( η * f[1] - f[0] )

plot( v, η )
xlabel( r'$\frac{v}{U_e}\sqrt{\mathrm{Re}_x}$' )
ylabel( r'$\eta$' )
grid()
xlim( (-0.1,1.1) )
ylim( (0,8) );
```

```python
dic = { '$\\eta$': η,
        '$f(\\eta)$': f[0],
        '$u/Ue$': f[1],
        '$f^{\\prime\\prime}$': f[2] }
df = pd.DataFrame( dic )
pd.set_option('display.max_rows', len(η) )
```

```python
latex_table = df.to_latex(
    index=False,
    escape=False,
    longtable=True,
    float_format="%.5f",
    column_format="cccc",
    caption="Blasius solution",
    label="tab:blasius"
)

latex_doc = f"""
\\documentclass{{article}}

\\usepackage{{amsmath}}
\\usepackage{{booktabs}}
\\usepackage{{longtable}}
\\usepackage{{geometry}}
\\geometry{{margin=2.5cm}}

\\begin{{document}}

\\section*{{Blasius Solution Table}}

{latex_table}

\\end{{document}}
"""

with open("blasius_table.tex", "w") as fout:
    fout.write(latex_doc)
```
