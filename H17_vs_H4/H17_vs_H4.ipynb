```python
η = np.linspace( 0.001, 0.999, 5001 )
```

```python
Ue4 = 2*η - 2*η**3 + η**4
func_θ4 = Ue4 * ( 1 - Ue4 )

Ue17 = η**(1/7)
func_θ17 = Ue17 * ( 1 - Ue17 )

plot( Ue4, η, label=r'$U^*_A=2\eta - 2\eta^3 + \eta^4$' )
plot( func_θ4, η, label=r'$U^*_A\left(1-U^*_A\right)$' )

plot( Ue17, η, label=r'$U^*_A=\eta^{1/7}$' )
plot( func_θ17, η, label=r'$U^*_B\left(1-U^*_B\right)$' )

title( r'$H_A=2.55, H_B=1.29$')
ylabel( r'$\eta$')
legend(loc='upper center')
grid()
xlim( (0,1) )
ylim( (0,1) );
```
